## Objetivo
Investigar y documentar diez funciones básicas de Python, explicar su utilidad y presentar ejemplos prácticos.

> Nota: En Python, algunas de estas herramientas son funciones incorporadas (*built-in functions*) y otras son funciones que permiten convertir datos. Todas se pueden utilizar directamente sin importar librerías adicionales.


## 1. `print()`

**¿Para qué sirve?** Muestra información en pantalla. Es útil para presentar mensajes, resultados o valores de variables.

**Ejemplo práctico:**

````python
nombre = "Bryan"
edad = 20
print("Hola", nombre)
print("Edad:", edad)
````

**Resultado esperado:**
```text
Hola Bryan
Edad: 20
```

## 2. `input()`

**¿Para qué sirve?** Permite solicitar información al usuario desde el teclado. El dato recibido se guarda como texto (`str`).

**Ejemplo práctico:**

````python
nombre = input("¿Cómo te llamas? ")
print("Bienvenido,", nombre)
````

**Resultado esperado:**
```text
Si el usuario escribe Ana:
Bienvenido, Ana
```

## 3. `len()`

**¿Para qué sirve?** Devuelve la cantidad de elementos de un objeto, por ejemplo, caracteres de un texto o elementos de una lista.

**Ejemplo práctico:**

````python
palabra = "Python"
print(len(palabra))

numeros = [10, 20, 30, 40]
print(len(numeros))
````

**Resultado esperado:**
```text
6
4
```

## 4. `type()`

**¿Para qué sirve?** Indica el tipo de dato de una variable, como entero (`int`), decimal (`float`) o texto (`str`).

**Ejemplo práctico:**

````python
edad = 18
precio = 12.5
nombre = "Laura"

print(type(edad))
print(type(precio))
print(type(nombre))
````

**Resultado esperado:**
```text
<class 'int'>
<class 'float'>
<class 'str'>
```

## 5. `int()`

**¿Para qué sirve?** Convierte un valor compatible a un número entero. Si convierte un decimal, elimina la parte decimal; no redondea.

**Ejemplo práctico:**

````python
numero = int("25")
print(numero)
print(type(numero))
````

**Resultado esperado:**
```text
25
<class 'int'>
```

## 6. `float()`

**¿Para qué sirve?** Convierte un valor compatible a un número decimal (`float`). Es útil para trabajar con medidas, precios o promedios.

**Ejemplo práctico:**

````python
altura = float("1.75")
print(altura)
print(type(altura))
````

**Resultado esperado:**
```text
1.75
<class 'float'>
```

## 7. `str()`

**¿Para qué sirve?** Convierte un valor a texto. Sirve para unir números con mensajes u otros textos.

**Ejemplo práctico:**

````python
edad = 20
mensaje = "Tengo " + str(edad) + " años"
print(mensaje)
````

**Resultado esperado:**
```text
Tengo 20 años
```

## 8. `range()`

**¿Para qué sirve?** Genera una secuencia de números, normalmente utilizada en ciclos `for`. El valor final no se incluye.

**Ejemplo práctico:**

````python
for numero in range(1, 6):
    print(numero)
````

**Resultado esperado:**
```text
1
2
3
4
5
```

## 9. `sum()`

**¿Para qué sirve?** Suma los elementos de una colección numérica, como una lista.

**Ejemplo práctico:**

````python
calificaciones = [4.0, 3.5, 4.5, 5.0]
total = sum(calificaciones)
print(total)
````

**Resultado esperado:**
```text
17.0
```

## 10. `max()`

**¿Para qué sirve?** Devuelve el valor más grande de una colección o de varios valores comparables.

**Ejemplo práctico:**

````python
temperaturas = [18, 22, 16, 25, 20]
print(max(temperaturas))
````

**Resultado esperado:**
```text
25
```

## Conclusión

Estas diez funciones permiten mostrar información, recibir datos, conocer sus características, convertir tipos de datos y realizar operaciones sencillas. Son herramientas fundamentales para comenzar a programar en Python y pueden combinarse para desarrollar programas más completos.
